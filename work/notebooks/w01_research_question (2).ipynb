## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
from google.colab import drive
import pandas as pd

drive.mount('/content/drive')

df = pd.read_csv(
    '/content/drive/MyDrive/flyrank-ml-internship/content_refresh_anonymized.csv'
)

# Quick look at the data

print(f"Total pages: {len(df):,}")

down_pages = (df["trend_direction"] == "down").sum()
down_pct = (df["trend_direction"] == "down").mean() * 100

print(f"Pages trending down: {down_pages:,} ({down_pct:.1f}%)")

old_pages = (df["days_since_last_update"] > 180).sum()
old_pct = (df["days_since_last_update"] > 180).mean() * 100

print(f"Pages not updated for more than 180 days: {old_pages:,} ({old_pct:.1f}%)")
```
